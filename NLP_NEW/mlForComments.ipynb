```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.decomposition import PCA
import nltk
from collections import Counter
from gensim.models import CoherenceModel, LdaModel
from gensim.corpora.dictionary import Dictionary

# Gerekli NLTK verilerini indiriyoruz (ilk çalıştırmada)
nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("averaged_perceptron_tagger", quiet=True)

#######################################
# Yardımcı Fonksiyonlar
#######################################

def compute_optimal_topics(tokenized_texts, start=2, limit=10, step=1):
    """
    Belirli tokenized_texts (örn. bir otelin yorumları için token listeleri) üzerinde,
    farklı konu sayılarını deneyip en yüksek coherence skoruna sahip LDA modelini seçer.
    """
    dictionary = Dictionary(tokenized_texts)
    corpus = [dictionary.doc2bow(text) for text in tokenized_texts]
    coherence_values = []
    model_list = []
    for num_topics in range(start, limit, step):
        model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, 
                         random_state=42, passes=10)
        model_list.append(model)
        coherencemodel = CoherenceModel(model=model, texts=tokenized_texts, dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())
    optimal_idx = np.argmax(coherence_values)
    optimal_topics = start + optimal_idx * step
    print(f"Optimal konu sayısı: {optimal_topics}")
    return optimal_topics, model_list[optimal_idx]

def compute_threshold(topic_weights, quantile=0.5):
    """
    Verilen konu ağırlıkları dağılımına göre (örneğin, ilk yorumun dağılımı) 
    belirli bir quantile'e göre threshold değeri hesaplar.
    """
    all_weights = np.array(topic_weights)
    threshold_value = np.quantile(all_weights, quantile)
    print(f"Threshold (quantile={quantile}): {threshold_value:.4f}")
    return threshold_value

def print_top_pca_terms(pca_components, feature_names, top_n=8):
    """
    Her PCA bileşeni için, bileşendeki terimlerin (TF-IDF ağırlıkları üzerinden) en baskın olanlarını yazdırır.
    """
    for i, comp in enumerate(pca_components):
        print(f"\nPCA Bileşeni {i} için en baskın {top_n} terim:")
        indices = np.argsort(np.abs(comp))[-top_n:]
        sorted_indices = indices[np.argsort(-np.abs(comp[indices]))]
        for idx in sorted_indices:
            term = feature_names[idx]
            weight = comp[idx]
            print(f"  {term:20} weight={weight:.4f}")

def pca_for_hotel(hotel_reviews):
    """
    Bir otelin yorumlarına TF-IDF uygulayarak PCA gerçekleştirir.
    Çıktıda, TF-IDF matrisinin boyutu, PCA açıklanan varyans oranları ve en baskın terimler verilir.
    Ayrıca PCA sonuçlarını 2B scatter plot olarak çizdirir.
    """
    tfidf_vectorizer = TfidfVectorizer(token_pattern=r'\S+')
    X_tfidf = tfidf_vectorizer.fit_transform(hotel_reviews)
    print("TF-IDF Matrisi Boyutu:", X_tfidf.shape)
    pca = PCA(n_components=2, random_state=42)
    X_pca = pca.fit_transform(X_tfidf.toarray())
    print("PCA açıklanan varyans oranları:", pca.explained_variance_ratio_)
    feature_names = tfidf_vectorizer.get_feature_names_out()
    print_top_pca_terms(pca.components_, feature_names, top_n=8)
    plt.figure(figsize=(6, 5))
    plt.scatter(X_pca[:, 0], X_pca[:, 1], color="blue")
    plt.title("PCA Görselleştirmesi")
    plt.xlabel("Birinci Ana Bileşen (PC1)")
    plt.ylabel("İkinci Ana Bileşen (PC2)")
    plt.show()
    return X_pca, pca.explained_variance_ratio_

def print_lda_topics(lda_model, n_top_words=10):
    """
    Gensim LDA modelinden bulunan konuları ve en baskın kelimeleri yazdırır.
    """
    topics = lda_model.show_topics(num_topics=-1, num_words=n_top_words, formatted=False)
    for topic_no, topic in topics:
        top_terms = [term for term, weight in topic]
        print(f"\nTopic {topic_no}:")
        print(" ".join(top_terms))

def lda_for_hotel(hotel_reviews, n_topics):
    """
    Bir otelin yorumları için Count Vectorizer ile vektörleştirme yapıp, Gensim LDA modelini uygular.
    Önce yorumlar tokenize edilir, sonra LDA modeli eğitilir ve konular yazdırılır.
    Ayrıca, ilk yorumun konu dağılımı üzerinden threshold değeri hesaplanır.
    """
    tokenized_reviews = [nltk.word_tokenize(review) for review in hotel_reviews]
    dictionary = Dictionary(tokenized_reviews)
    corpus = [dictionary.doc2bow(text) for text in tokenized_reviews]
    lda_model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=n_topics, random_state=42, passes=10)
    print_lda_topics(lda_model, n_top_words=8)
    # Örnek: İlk yorum için konu dağılımını alıp threshold hesaplayalım
    topic_distribution = lda_model.get_document_topics(corpus[0])
    topic_weights = [weight for topic_id, weight in topic_distribution]
    threshold = compute_threshold(topic_weights, quantile=0.5)
    return lda_model, threshold

#######################################
# Ana Program: İlk 20 Otel için PCA ve LDA
#######################################

if __name__ == "__main__":
    # CSV verisini oku
    df = pd.read_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\final_processed_reviews_deepl_use.csv", encoding="utf-8")
    # "hotel_name" ve "processed_final_review" sütunlarının boş olmamasını sağla
    df.dropna(subset=["hotel_name", "processed_final_review"], inplace=True)
    
    # İlk 20 oteli seç
    hotel_names = df["hotel_name"].unique()[:20]
    
    for hotel in hotel_names:
        print(f"=== {hotel} ===")
        hotel_reviews = df.loc[df["hotel_name"] == hotel, "processed_final_review"].tolist()
        print(f"{hotel} için yorum sayısı: {len(hotel_reviews)}")
        if len(hotel_reviews) < 2:
            print("Yeterli yorum olmadığı için analiz yapılamıyor. Devam ediliyor...\n")
            continue
        
        print("\n--- PCA Analizi ---")
        X_pca, var_ratios = pca_for_hotel(hotel_reviews)
        
        print("\n--- LDA Analizi ---")
        # Önce, yorumları tokenize et
        tokenized_reviews = [nltk.word_tokenize(review) for review in hotel_reviews]
        # Optimal konu sayısını belirleyelim
        optimal_topics, lda_model_opt = compute_optimal_topics(tokenized_reviews, start=2, limit=10, step=1)
        print("\nLDA ile çıkan konular:")
        print_lda_topics(lda_model_opt, n_top_words=8)
        
        # Threshold hesaplaması (örneğin, ilk yorum üzerinden)
        dictionary = Dictionary(tokenized_reviews)
        corpus = [dictionary.doc2bow(text) for text in tokenized_reviews]
        topic_distribution = lda_model_opt.get_document_topics(corpus[0])
        topic_weights = [weight for topic_id, weight in topic_distribution]
        threshold_value = compute_threshold(topic_weights, quantile=0.5)
        
        print("\n" + "-"*50 + "\n")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.decomposition import PCA
import nltk
from collections import Counter
from gensim.models import CoherenceModel, LdaModel
from gensim.corpora.dictionary import Dictionary
from transformers import pipeline

# Gerekli NLTK verilerini indiriyoruz
nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("averaged_perceptron_tagger", quiet=True)

##########################################
# Fonksiyonlar
##########################################

def print_top_pca_terms(pca_components, feature_names, top_n=8):
    for i, comp in enumerate(pca_components):
        print(f"\nPCA Bileşeni {i} için en baskın {top_n} terim:")
        indices = np.argsort(np.abs(comp))[-top_n:]
        sorted_indices = indices[np.argsort(-np.abs(comp[indices]))]
        for idx in sorted_indices:
            term = feature_names[idx]
            weight = comp[idx]
            print(f"  {term:20} weight={weight:.4f}")

def compute_optimal_topics(tokenized_texts, start=2, limit=10, step=1):
    dictionary = Dictionary(tokenized_texts)
    corpus = [dictionary.doc2bow(text) for text in tokenized_texts]
    coherence_values = []
    model_list = []
    for num_topics in range(start, limit, step):
        model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, 
                         random_state=42, passes=10)
        model_list.append(model)
        coherencemodel = CoherenceModel(model=model, texts=tokenized_texts, 
                                         dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())
    optimal_idx = np.argmax(coherence_values)
    optimal_topics = start + optimal_idx * step
    print(f"Optimal konu sayısı: {optimal_topics}")
    return optimal_topics, model_list[optimal_idx], dictionary, corpus

def compute_threshold_for_hotel(lda_model, corpus, n_topics, quantile=0.5):
    all_topic_weights = []
    for bow in corpus:
        doc_topic = lda_model.get_document_topics(bow)
        vec = np.zeros(n_topics)
        for topic_id, weight in doc_topic:
            vec[topic_id] = weight
        all_topic_weights.append(vec)
    all_topic_weights = np.array(all_topic_weights)
    avg_topic_distribution = np.mean(all_topic_weights, axis=0)
    threshold_value = np.quantile(avg_topic_distribution, quantile)
    print(f"Threshold (quantile={quantile}): {threshold_value:.4f}")
    return threshold_value, avg_topic_distribution

def pca_for_hotel(hotel_reviews):
    tfidf_vectorizer = TfidfVectorizer(token_pattern=r'\S+')
    X_tfidf = tfidf_vectorizer.fit_transform(hotel_reviews)
    print("TF-IDF Matrisi Boyutu:", X_tfidf.shape)
    pca = PCA(n_components=2, random_state=42)
    X_pca = pca.fit_transform(X_tfidf.toarray())
    print("PCA açıklanan varyans oranları:", pca.explained_variance_ratio_)
    feature_names = tfidf_vectorizer.get_feature_names_out()
    print_top_pca_terms(pca.components_, feature_names, top_n=8)
    plt.figure(figsize=(6, 5))
    plt.scatter(X_pca[:, 0], X_pca[:, 1], color="blue")
    plt.title("PCA Görselleştirmesi")
    plt.xlabel("Birinci Ana Bileşen (PC1)")
    plt.ylabel("İkinci Ana Bileşen (PC2)")
    plt.show()
    avg_pca = np.mean(X_pca, axis=0)
    return avg_pca, pca.explained_variance_ratio_

def print_lda_topics(lda_model, n_top_words=8):
    topics = lda_model.show_topics(num_topics=-1, num_words=n_top_words, formatted=False)
    topic_dict = {}
    for topic_no, topic in topics:
        top_terms = [term for term, weight in topic]
        topic_dict[topic_no] = top_terms
        print(f"\nTopic {topic_no}:")
        print(" ".join(top_terms))
    return topic_dict

def lda_for_hotel(hotel_reviews, fixed_n_topics):
    tokenized_reviews = [nltk.word_tokenize(review) for review in hotel_reviews]
    optimal_topics, lda_model, dictionary, corpus = compute_optimal_topics(tokenized_reviews, start=2, limit=fixed_n_topics+1, step=1)
    print("\nLDA ile çıkan konular:")
    topic_dict = print_lda_topics(lda_model, n_top_words=8)
    n_topics = optimal_topics  # sabit konu sayısı
    lda_features = np.zeros((len(corpus), n_topics))
    for i, bow in enumerate(corpus):
        for topic_id, weight in lda_model.get_document_topics(bow):
            lda_features[i, topic_id] = weight
    avg_topic_distribution = np.mean(lda_features, axis=0)
    print("\nOrtalama LDA konu dağılımı:", avg_topic_distribution)
    threshold = np.quantile(avg_topic_distribution, 0.5)
    print(f"Threshold (quantile=0.5): {threshold:.4f}")
    binary_lda = (avg_topic_distribution >= threshold).astype(int)
    print("Binary LDA vektörü:", binary_lda)
    return avg_topic_distribution, binary_lda, n_topics, lda_model, corpus, topic_dict

# Duygu Analizi için BERT tabanlı sentiment pipeline kurulumu
from transformers import pipeline
sentiment_pipeline = pipeline("sentiment-analysis", model="cardiffnlp/twitter-roberta-base-sentiment", tokenizer="cardiffnlp/twitter-roberta-base-sentiment")

def get_sentiment(review_text):
    result = sentiment_pipeline(review_text)[0]
    label = result['label']
    if label.lower() == "negative":
        return -1
    elif label.lower() == "neutral":
        return 0
    elif label.lower() == "positive":
        return 1
    else:
        return 0

def aspect_level_sentiment(hotel_reviews, lda_model, corpus, n_topics, threshold):
    aspect_sentiments = []
    # Her konu için ayrı ayrı
    for topic_idx in range(n_topics):
        aspect_reviews = []
        for i, bow in enumerate(corpus):
            topic_weight = 0
            for t_id, weight in lda_model.get_document_topics(bow):
                if t_id == topic_idx:
                    topic_weight = weight
                    break
            if topic_weight >= threshold:
                aspect_reviews.append(hotel_reviews[i])
        if aspect_reviews:
            sentiments = [get_sentiment(review) for review in aspect_reviews]
            avg_sent = np.mean(sentiments)
            if avg_sent > 0.2:
                final_sent = 1
            elif avg_sent < -0.2:
                final_sent = -1
            else:
                final_sent = 0
        else:
            final_sent = 0
        aspect_sentiments.append(final_sent)
    return np.array(aspect_sentiments)

##########################################
# Global Analiz: Tüm oteller için özellik vektörlerini hesaplama
##########################################

if __name__ == "__main__":
    # CSV verisini oku
    df = pd.read_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\final_processed_reviews_deepl_use.csv", encoding="utf-8")
    df.dropna(subset=["hotel_name", "processed_final_review"], inplace=True)
    hotels = df["hotel_name"].unique()
    
    global_features = []  # Her otelin özellik sözlüğü
    fixed_n_topics = 7  # Sabit konu sayısı (örneğin 7)
    
    for hotel in hotels:
        print(f"\n=== {hotel} ===")
        hotel_reviews = df.loc[df["hotel_name"] == hotel, "processed_final_review"].tolist()
        print(f"{hotel} için yorum sayısı: {len(hotel_reviews)}")
        if len(hotel_reviews) < 2:
            print("Yeterli yorum yok, atlanıyor.")
            continue
        
        # Opsiyonel: PCA analizi (görselleştirme)
        print("\n--- PCA Analizi ---")
        avg_pca, var_ratios = pca_for_hotel(hotel_reviews)
        
        print("\n--- LDA Analizi ---")
        avg_lda, binary_lda, n_topics, lda_model, corpus, topic_dict = lda_for_hotel(hotel_reviews, fixed_n_topics)
        
        # Aspect-level sentiment analizi: Tüm yorumlar üzerinden threshold değeri avg_topic_distribution hesaplanıyor
        threshold_value, avg_topic_distribution = compute_threshold_for_hotel(lda_model, corpus, n_topics, quantile=0.5)
        aspect_sentiments = aspect_level_sentiment(hotel_reviews, lda_model, corpus, n_topics, threshold_value)
        
        # Global özellik vektörü oluşturuluyor (sabit sütun isimleri)
        feature_dict = {"hotel_name": hotel}
        for i in range(fixed_n_topics):
            # Column isimleri; aynı zamanda hangi konu terimlerinin (aspect) temsil ettiğini de ekrana bastırıyoruz.
            topic_words = " ".join(topic_dict.get(i, []))
            feature_dict[f"aspect_topic_{i}"] = int(binary_lda[i])
            feature_dict[f"sentiment_topic_{i}"] = int(aspect_sentiments[i])
            print(f"Column 'aspect_topic_{i}' ({topic_words}): Binary = {binary_lda[i]}, Sentiment = {aspect_sentiments[i]}")
        global_features.append(feature_dict)
        
        print("\n" + "-"*50)
    
    # Global özellikleri DataFrame olarak göster
    features_df = pd.DataFrame(global_features)
    print("\n***** Global Özellik Matrisi *****")
    print(features_df)
    
    # CSV'ye kaydetme (isteğe bağlı, ayrı kodla yapılabilir)

    features_df.to_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\hotel_aspect_sentiment_features.csv", index=False, encoding="utf-8-sig")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.decomposition import PCA
import nltk
from collections import Counter
from gensim.models import CoherenceModel, LdaModel
from gensim.corpora.dictionary import Dictionary
import torch
from transformers import AutoTokenizer, AutoModel

# Gerekli NLTK verilerini indiriyoruz
nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("averaged_perceptron_tagger", quiet=True)

##########################################
# Fonksiyonlar
##########################################

def print_top_pca_terms(pca_components, feature_names, top_n=8):
    for i, comp in enumerate(pca_components):
        print(f"\nPCA Bileşeni {i} için en baskın {top_n} terim:")
        indices = np.argsort(np.abs(comp))[-top_n:]
        sorted_indices = indices[np.argsort(-np.abs(comp[indices]))]
        for idx in sorted_indices:
            term = feature_names[idx]
            weight = comp[idx]
            print(f"  {term:20} weight={weight:.4f}")

def compute_optimal_topics(tokenized_texts, start=2, limit=10, step=1):
    dictionary = Dictionary(tokenized_texts)
    corpus = [dictionary.doc2bow(text) for text in tokenized_texts]
    coherence_values = []
    model_list = []
    for num_topics in range(start, limit, step):
        model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, 
                         random_state=42, passes=10)
        model_list.append(model)
        coherencemodel = CoherenceModel(model=model, texts=tokenized_texts, 
                                         dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())
    optimal_idx = np.argmax(coherence_values)
    optimal_topics = start + optimal_idx * step
    print(f"Optimal konu sayısı: {optimal_topics}")
    return optimal_topics, model_list[optimal_idx], dictionary, corpus


def print_lda_topics(lda_model, n_top_words=8):
    """ LDA modelindeki her topic için en baskın kelimeleri döndürür. """
    topics = lda_model.show_topics(num_topics=-1, num_words=n_top_words, formatted=False)
    topic_dict = {}
    for topic_no, topic in topics:
        top_terms = [term for term, _ in topic]
        topic_dict[topic_no] = top_terms
    return topic_dict

def compute_threshold_for_hotel(lda_model, corpus, n_topics, quantile=0.5):
    all_topic_weights = []
    for bow in corpus:
        doc_topic = lda_model.get_document_topics(bow)
        vec = np.zeros(n_topics)
        for topic_id, weight in doc_topic:
            vec[topic_id] = weight
        all_topic_weights.append(vec)
    all_topic_weights = np.array(all_topic_weights)
    avg_topic_distribution = np.mean(all_topic_weights, axis=0)
    threshold_value = np.quantile(avg_topic_distribution, quantile)
    print(f"Threshold (quantile={quantile}): {threshold_value:.4f}")
    return threshold_value, avg_topic_distribution

def pca_for_hotel(hotel_reviews):
    tfidf_vectorizer = TfidfVectorizer(token_pattern=r'\S+')
    X_tfidf = tfidf_vectorizer.fit_transform(hotel_reviews)
    print("TF-IDF Matrisi Boyutu:", X_tfidf.shape)
    pca = PCA(n_components=2, random_state=42)
    X_pca = pca.fit_transform(X_tfidf.toarray())
    print("PCA açıklanan varyans oranları:", pca.explained_variance_ratio_)
    feature_names = tfidf_vectorizer.get_feature_names_out()
    print_top_pca_terms(pca.components_, feature_names, top_n=8)
    plt.figure(figsize=(6, 5))
    plt.scatter(X_pca[:, 0], X_pca[:, 1], color="blue")
    plt.title("PCA Görselleştirmesi")
    plt.xlabel("Birinci Ana Bileşen (PC1)")
    plt.ylabel("İkinci Ana Bileşen (PC2)")
    plt.show()
    avg_pca = np.mean(X_pca, axis=0)
    return avg_pca, pca.explained_variance_ratio_

def print_lda_topics(lda_model, n_top_words=8):
    topics = lda_model.show_topics(num_topics=-1, num_words=n_top_words, formatted=False)
    topic_dict = {}
    for topic_no, topic in topics:
        top_terms = [term for term, weight in topic]
        topic_dict[topic_no] = top_terms
        print(f"\nTopic {topic_no}:")
        print(" ".join(top_terms))
    return topic_dict

def lda_for_hotel(hotel_reviews, fixed_n_topics):
    tokenized_reviews = [nltk.word_tokenize(review) for review in hotel_reviews]
    optimal_topics, lda_model, dictionary, corpus = compute_optimal_topics(tokenized_reviews, start=2, limit=fixed_n_topics+1, step=1)
    print("\nLDA ile çıkan konular:")
    topic_dict = print_lda_topics(lda_model, n_top_words=8)
    n_topics = optimal_topics  # Bu, o otel için belirlenen optimal konu sayısıdır.
    lda_features = np.zeros((len(corpus), n_topics))
    for i, bow in enumerate(corpus):
        for topic_id, weight in lda_model.get_document_topics(bow):
            lda_features[i, topic_id] = weight
    avg_topic_distribution = np.mean(lda_features, axis=0)
    print("\nOrtalama LDA konu dağılımı:", avg_topic_distribution)
    threshold = np.quantile(avg_topic_distribution, 0.5)
    print(f"Threshold (quantile=0.5): {threshold:.4f}")
    binary_lda = (avg_topic_distribution >= threshold).astype(int)
    print("Binary LDA vektörü:", binary_lda)
    return avg_topic_distribution, binary_lda, n_topics, lda_model, corpus, topic_dict

##########################################
# Hugging Face BERT Entegrasyonu
##########################################
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
bert_model = AutoModel.from_pretrained("bert-base-uncased")
bert_model.eval()

def get_bert_embedding(text):
    inputs = bert_tokenizer(text, return_tensors="pt", truncation=True, max_length=512)
    with torch.no_grad():
        outputs = bert_model(**inputs)
    cls_embedding = outputs.last_hidden_state[:, 0, :]
    return cls_embedding.detach().numpy()

def bert_for_hotel(hotel_reviews):
    embeddings = []
    for review in hotel_reviews:
        emb = get_bert_embedding(review)
        embeddings.append(emb)
    avg_embedding = np.mean(np.vstack(embeddings), axis=0)
    print("Average BERT embedding shape:", avg_embedding.shape)
    return avg_embedding

##########################################
# Global Analiz & CSV’ye Kaydetme
##########################################
if __name__ == "__main__":
    # CSV verisini oku (sütun adları: "hotel_name", "processed_final_review")
    df = pd.read_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\final_processed_reviews_deepl_use.csv", encoding="utf-8")
    df.dropna(subset=["hotel_name", "processed_final_review"], inplace=True)
    hotels = df["hotel_name"].unique()
    
    results = []
    fixed_n_topics = 7  # Global olarak kullanmak istediğimiz sabit konu sayısı
    # Bu değere göre, tüm oteller için final feature vektörleri fixed_n_topics kadar aspect sütunu içerecek.
    
    for hotel in hotels:
        print(f"\n=== {hotel} ===")
        hotel_reviews = df.loc[df["hotel_name"] == hotel, "processed_final_review"].tolist()
        print(f"{hotel} için yorum sayısı: {len(hotel_reviews)}")
        if len(hotel_reviews) < 2:
            print("Yeterli yorum yok, atlanıyor.")
            continue
        
        # PCA analizi (sadece görselleştirme ve özet)
        print("\n--- PCA Analizi ---")
        avg_pca, var_ratios = pca_for_hotel(hotel_reviews)
        
        # LDA analizi
        print("\n--- LDA Analizi ---")
        avg_lda, binary_lda, n_topics, lda_model, corpus, topic_dict = lda_for_hotel(hotel_reviews, fixed_n_topics)
        # Global özellikler sabit sütun sayısı için:
        # Eğer n_topics (o otel için optimal konu sayısı) fixed_n_topics'den az ise, pad ile sıfır ekleyelim.
        if n_topics < fixed_n_topics:
            pad_length = fixed_n_topics - n_topics
            binary_lda = np.pad(binary_lda, (0, pad_length), mode='constant')
            for i in range(n_topics, fixed_n_topics):
                topic_dict[i] = ["N/A"]
        elif n_topics > fixed_n_topics:
            # Eğer n_topics daha fazla ise, sadece ilk fixed_n_topics konuyu kullanıyoruz.
            binary_lda = binary_lda[:fixed_n_topics]
            topic_dict = {i: topic_dict[i] for i in range(fixed_n_topics)}
        
        # Threshold: Otelin tüm yorumlarının konu dağılımından hesaplanıyor (compute_threshold_for_hotel)
        threshold_value, avg_topic_distribution = compute_threshold_for_hotel(lda_model, corpus, n_topics, quantile=0.5)
        
        # BERT embedding analizi
        print("\n--- BERT Embedding Analizi ---")
        avg_bert = bert_for_hotel(hotel_reviews)
        bert_str = ",".join(map(str, avg_bert.tolist()))
        
        # Sonuçları dictionary olarak sakla
        hotel_result = {
            "hotel_name": hotel,
            "pca_pc1": avg_pca[0],
            "pca_pc2": avg_pca[1],
            "bert_embedding": bert_str
        }
        for i in range(fixed_n_topics):
            topic_words = " ".join(topic_dict.get(i, ["N/A"]))
            hotel_result[f"aspect_topic_{i}"] = int(binary_lda[i])
            # Sentiment analizi için placeholder: ileride BERT tabanlı aspect-level sentiment analizi eklenecek.
            hotel_result[f"sentiment_topic_{i}"] = 0  
            print(f"Column 'aspect_topic_{i}' ({topic_words}): Binary = {binary_lda[i]}, Sentiment = {0}")
        
        results.append(hotel_result)
        print("\n" + "-"*50)
    
    results_df = pd.DataFrame(results)
    results_df.to_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\hotel_feature_matrix.csv", index=False, encoding="utf-8-sig")
    print("\n***** Global Sonuçlar CSV'ye kaydedildi: hotel_feature_matrix.csv *****")
```

```python
 import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.decomposition import PCA
import nltk
from collections import Counter
from gensim.models import CoherenceModel, LdaModel
from gensim.corpora.dictionary import Dictionary
import torch
from transformers import AutoTokenizer, AutoModel

# Gerekli NLTK verilerini indiriyoruz
nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("averaged_perceptron_tagger", quiet=True)

##########################################
# Fonksiyonlar
##########################################

def print_top_pca_terms(pca_components, feature_names, top_n=8):
    for i, comp in enumerate(pca_components):
        print(f"\nPCA Bileşeni {i} için en baskın {top_n} terim:")
        indices = np.argsort(np.abs(comp))[-top_n:]
        sorted_indices = indices[np.argsort(-np.abs(comp[indices]))]
        for idx in sorted_indices:
            term = feature_names[idx]
            weight = comp[idx]
            print(f"  {term:20} weight={weight:.4f}")

def compute_optimal_topics(tokenized_texts, start=2, limit=10, step=1):
    dictionary = Dictionary(tokenized_texts)
    corpus = [dictionary.doc2bow(text) for text in tokenized_texts]
    coherence_values = []
    model_list = []
    for num_topics in range(start, limit, step):
        model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, 
                         random_state=42, passes=10)
        model_list.append(model)
        coherencemodel = CoherenceModel(model=model, texts=tokenized_texts, 
                                         dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())
    optimal_idx = np.argmax(coherence_values)
    optimal_topics = start + optimal_idx * step
    print(f"Optimal konu sayısı: {optimal_topics}")
    return optimal_topics, model_list[optimal_idx], dictionary, corpus

def compute_threshold_for_hotel(lda_model, corpus, n_topics, quantile=0.5):
    all_topic_weights = []
    for bow in corpus:
        doc_topic = lda_model.get_document_topics(bow)
        vec = np.zeros(n_topics)
        for topic_id, weight in doc_topic:
            vec[topic_id] = weight
        all_topic_weights.append(vec)
    all_topic_weights = np.array(all_topic_weights)
    avg_topic_distribution = np.mean(all_topic_weights, axis=0)
    threshold_value = np.quantile(avg_topic_distribution, quantile)
    print(f"Threshold (quantile={quantile}): {threshold_value:.4f}")
    return threshold_value, avg_topic_distribution

def pca_for_hotel(hotel_reviews):
    tfidf_vectorizer = TfidfVectorizer(token_pattern=r'\S+')
    X_tfidf = tfidf_vectorizer.fit_transform(hotel_reviews)
    print("TF-IDF Matrisi Boyutu:", X_tfidf.shape)
    pca = PCA(n_components=2, random_state=42)
    X_pca = pca.fit_transform(X_tfidf.toarray())
    print("PCA açıklanan varyans oranları:", pca.explained_variance_ratio_)
    feature_names = tfidf_vectorizer.get_feature_names_out()
    print_top_pca_terms(pca.components_, feature_names, top_n=8)
    plt.figure(figsize=(6, 5))
    plt.scatter(X_pca[:, 0], X_pca[:, 1], color="blue")
    plt.title("PCA Görselleştirmesi")
    plt.xlabel("Birinci Ana Bileşen (PC1)")
    plt.ylabel("İkinci Ana Bileşen (PC2)")
    plt.show()
    avg_pca = np.mean(X_pca, axis=0)
    return avg_pca, pca.explained_variance_ratio_

def print_lda_topics(lda_model, n_top_words=8):
    topics = lda_model.show_topics(num_topics=-1, num_words=n_top_words, formatted=False)
    topic_dict = {}
    for topic_no, topic in topics:
        top_terms = [term for term, weight in topic]
        topic_dict[topic_no] = top_terms
        print(f"\nTopic {topic_no}:")
        print(" ".join(top_terms))
    return topic_dict

def lda_for_hotel(hotel_reviews, fixed_n_topics):
    tokenized_reviews = [nltk.word_tokenize(review) for review in hotel_reviews]
    optimal_topics, lda_model, dictionary, corpus = compute_optimal_topics(tokenized_reviews, start=2, limit=fixed_n_topics+1, step=1)
    print("\nLDA ile çıkan konular:")
    topic_dict = print_lda_topics(lda_model, n_top_words=8)
    n_topics = optimal_topics  # Bu, o otel için belirlenen optimal konu sayısıdır.
    lda_features = np.zeros((len(corpus), n_topics))
    for i, bow in enumerate(corpus):
        for topic_id, weight in lda_model.get_document_topics(bow):
            lda_features[i, topic_id] = weight
    avg_topic_distribution = np.mean(lda_features, axis=0)
    print("\nOrtalama LDA konu dağılımı:", avg_topic_distribution)
    threshold = np.quantile(avg_topic_distribution, 0.5)
    print(f"Threshold (quantile=0.5): {threshold:.4f}")
    binary_lda = (avg_topic_distribution >= threshold).astype(int)
    print("Binary LDA vektörü:", binary_lda)
    return avg_topic_distribution, binary_lda, n_topics, lda_model, corpus, topic_dict

##########################################
# Hugging Face BERT Entegrasyonu
##########################################
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
bert_model = AutoModel.from_pretrained("bert-base-uncased")
bert_model.eval()

def get_bert_embedding(text):
    inputs = bert_tokenizer(text, return_tensors="pt", truncation=True, max_length=512)
    with torch.no_grad():
        outputs = bert_model(**inputs)
    cls_embedding = outputs.last_hidden_state[:, 0, :]
    return cls_embedding.detach().numpy()

def bert_for_hotel(hotel_reviews):
    embeddings = []
    for review in hotel_reviews:
        emb = get_bert_embedding(review)
        embeddings.append(emb)
    avg_embedding = np.mean(np.vstack(embeddings), axis=0)
    print("Average BERT embedding shape:", avg_embedding.shape)
    return avg_embedding

##########################################
# Global Analiz & CSV’ye Kaydetme
##########################################
if __name__ == "__main__":
    # CSV verisini oku (sütun adları: "hotel_name", "processed_final_review")
    df = pd.read_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\final_processed_reviews_deepl_use.csv", encoding="utf-8")
    df.dropna(subset=["hotel_name", "processed_final_review"], inplace=True)
    hotels = df["hotel_name"].unique()
    
    results = []
    fixed_n_topics = 7  # Global olarak kullanmak istediğimiz sabit konu sayısı
    # Bu değere göre, tüm oteller için final feature vektörleri fixed_n_topics kadar aspect sütunu içerecek.
    
    for hotel in hotels:
        print(f"\n=== {hotel} ===")
        hotel_reviews = df.loc[df["hotel_name"] == hotel, "processed_final_review"].tolist()
        print(f"{hotel} için yorum sayısı: {len(hotel_reviews)}")
        if len(hotel_reviews) < 2:
            print("Yeterli yorum yok, atlanıyor.")
            continue
        
        # PCA analizi (sadece görselleştirme ve özet)
        print("\n--- PCA Analizi ---")
        avg_pca, var_ratios = pca_for_hotel(hotel_reviews)
        
        # LDA analizi
        print("\n--- LDA Analizi ---")
        avg_lda, binary_lda, n_topics, lda_model, corpus, topic_dict = lda_for_hotel(hotel_reviews, fixed_n_topics)
        # Global özellikler sabit sütun sayısı için:
        # Eğer n_topics (o otel için optimal konu sayısı) fixed_n_topics'den az ise, pad ile sıfır ekleyelim.
        if n_topics < fixed_n_topics:
            pad_length = fixed_n_topics - n_topics
            binary_lda = np.pad(binary_lda, (0, pad_length), mode='constant')
            for i in range(n_topics, fixed_n_topics):
                topic_dict[i] = ["N/A"]
        elif n_topics > fixed_n_topics:
            # Eğer n_topics daha fazla ise, sadece ilk fixed_n_topics konuyu kullanıyoruz.
            binary_lda = binary_lda[:fixed_n_topics]
            topic_dict = {i: topic_dict[i] for i in range(fixed_n_topics)}
        
        # Threshold: Otelin tüm yorumlarının konu dağılımından hesaplanıyor (compute_threshold_for_hotel)
        threshold_value, avg_topic_distribution = compute_threshold_for_hotel(lda_model, corpus, n_topics, quantile=0.5)
        
        # BERT embedding analizi
        print("\n--- BERT Embedding Analizi ---")
        avg_bert = bert_for_hotel(hotel_reviews)
        bert_str = ",".join(map(str, avg_bert.tolist()))
        
        # Sonuçları dictionary olarak sakla
        hotel_result = {
            "hotel_name": hotel,
            "pca_pc1": avg_pca[0],
            "pca_pc2": avg_pca[1],
            "bert_embedding": bert_str
        }
        for i in range(fixed_n_topics):
            topic_words = " ".join(topic_dict.get(i, ["N/A"]))
            hotel_result[f"aspect_topic_{i}"] = int(binary_lda[i])
            # Sentiment analizi için placeholder: ileride BERT tabanlı aspect-level sentiment analizi eklenecek.
            hotel_result[f"sentiment_topic_{i}"] = 0  
            print(f"Column 'aspect_topic_{i}' ({topic_words}): Binary = {binary_lda[i]}, Sentiment = {0}")
        
        results.append(hotel_result)
        print("\n" + "-"*50)
    
    results_df = pd.DataFrame(results)
    results_df.to_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\hotel_feature_matrix.csv", index=False, encoding="utf-8-sig")
    print("\n***** Global Sonuçlar CSV'ye kaydedildi: hotel_feature_matrix.csv *****")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import nltk
from gensim.models import CoherenceModel, LdaModel
from gensim.corpora.dictionary import Dictionary

import torch
from transformers import AutoTokenizer, AutoModel, pipeline

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA

import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

# NLTK verileri
nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("averaged_perceptron_tagger", quiet=True)

def compute_optimal_topics(tokenized_texts, start=2, limit=10, step=1):
    """
    Farklı topic sayıları (start..limit) deneyerek en iyi coherence değeri olan modeli bulur.
    """
    dictionary = Dictionary(tokenized_texts)
    corpus = [dictionary.doc2bow(text) for text in tokenized_texts]
    coherence_values = []
    model_list = []
    for num_topics in range(start, limit, step):
        model = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics,
                         random_state=42, passes=10)
        model_list.append(model)
        coherencemodel = CoherenceModel(model=model, texts=tokenized_texts,
                                         dictionary=dictionary, coherence='c_v')
        coherence_values.append(coherencemodel.get_coherence())
    optimal_idx = np.argmax(coherence_values)
    optimal_topics = start + optimal_idx * step
    return optimal_topics, model_list[optimal_idx], dictionary, corpus

def lda_for_hotel(hotel_reviews, fixed_n_topics):
    """
    Otel bazında LDA analizi:
      - Yorumları tokenize
      - En iyi topic sayısını bul
      - threshold ile binary presence
    """
    tokenized_reviews = [nltk.word_tokenize(r) for r in hotel_reviews]
    optimal_topics, lda_model, dictionary, corpus = compute_optimal_topics(
        tokenized_reviews, start=2, limit=fixed_n_topics+1, step=1
    )
    
    # Topic -> kelime listesi
    topics = lda_model.show_topics(num_topics=-1, num_words=8, formatted=False)
    topic_dict = {}
    for topic_no, topic in topics:
        top_terms = [term for term, _ in topic]
        topic_dict[topic_no] = top_terms
    
    # Her doküman (yorum) için topic dağılımları
    n_topics = optimal_topics
    lda_features = np.zeros((len(corpus), n_topics))
    for i, bow in enumerate(corpus):
        for topic_id, weight in lda_model.get_document_topics(bow):
            lda_features[i, topic_id] = weight
    
    # Otel bazında ortalama topic dağılımı
    avg_topic_distribution = np.mean(lda_features, axis=0)
    # Eşik (median) => presence
    threshold = np.quantile(avg_topic_distribution, 0.5)
    binary_lda = (avg_topic_distribution >= threshold).astype(int)
    
    # Gerekirse pad / kes
    if n_topics < fixed_n_topics:
        pad_length = fixed_n_topics - n_topics
        binary_lda = np.pad(binary_lda, (0, pad_length), mode='constant')
        for i in range(n_topics, fixed_n_topics):
            topic_dict[i] = ["N/A"]
    elif n_topics > fixed_n_topics:
        binary_lda = binary_lda[:fixed_n_topics]
        topic_dict = {i: topic_dict[i] for i in range(fixed_n_topics)}
    
    return lda_model, corpus, topic_dict, binary_lda

# Sentiment pipeline (model & tokenizer)
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment",
    tokenizer="nlptown/bert-base-multilingual-uncased-sentiment"
)

def sentiment_to_score(label):
    """
    '1 star' -> -1
    '2 star' -> -1
    '3 star' ->  0
    '4 star' -> +1
    '5 star' -> +1
    """
    if label.startswith("1 star") or label.startswith("2 star"):
        return -1
    elif label.startswith("3 star"):
        return 0
    else:
        return 1

def compute_aspect_sentiment_for_hotel(lda_model, corpus, n_topics, hotel_reviews, threshold_doc=0.2):
    """
    Her topic için, doc_topic_weight > threshold_doc olan yorumları al.
    Ardından sentiment_pipeline ile tahmin yap, ortalamasını -1, 0, +1'e yuvarla.
    """
    # doc-topic distribution
    doc_topic_weights = []
    for bow in corpus:
        topic_weights = lda_model.get_document_topics(bow)
        # topic_weights: [(topic_id, weight), ...]
        doc_topic_weights.append(dict(topic_weights))
    
    aspect_sentiments = []
    for topic_id in range(n_topics):
        # Bu topic'e ait yorumları bul
        relevant_reviews = []
        for doc_idx, weights_dict in enumerate(doc_topic_weights):
            weight = weights_dict.get(topic_id, 0.0)
            if weight >= threshold_doc:
                relevant_reviews.append(hotel_reviews[doc_idx])
        
        if len(relevant_reviews) == 0:
            # O topic'e ait yorum yoksa => 0
            aspect_sentiments.append(0)
            continue
        
        # Aksi halde, her review için sentiment alıp ortalamasını bulalım
        scores = []
        for rev in relevant_reviews:
            result = sentiment_pipeline(rev[:512])[0]  # long text varsa kısalt
            score = sentiment_to_score(result["label"])
            scores.append(score)
        
        avg_score = np.mean(scores)  # float
        # yuvarlama => -1, 0, +1
        if avg_score > 0.2:
            final_sentiment = 1
        elif avg_score < -0.2:
            final_sentiment = -1
        else:
            final_sentiment = 0
        
        aspect_sentiments.append(final_sentiment)
    
    return aspect_sentiments


def pca_for_hotel(hotel_reviews):
    tfidf_vectorizer = TfidfVectorizer(token_pattern=r'\S+')
    X_tfidf = tfidf_vectorizer.fit_transform(hotel_reviews)
    pca = PCA(n_components=2, random_state=42)
    X_pca = pca.fit_transform(X_tfidf.toarray())
    avg_pca = np.mean(X_pca, axis=0)
    return avg_pca, pca.explained_variance_ratio_

# BERT embedding (opsiyonel)
bert_tokenizer_base = AutoTokenizer.from_pretrained("bert-base-uncased")
bert_model_base = AutoModel.from_pretrained("bert-base-uncased")
bert_model_base.eval()

def get_bert_embedding(text):
    inputs = bert_tokenizer_base(text, return_tensors="pt", truncation=True, max_length=512)
    with torch.no_grad():
        outputs = bert_model_base(**inputs)
    cls_embedding = outputs.last_hidden_state[:, 0, :]
    return cls_embedding.detach().numpy()

def bert_for_hotel(hotel_reviews):
    embeddings = []
    for review in hotel_reviews:
        emb = get_bert_embedding(review)
        embeddings.append(emb)
    avg_embedding = np.mean(np.vstack(embeddings), axis=0)
    return avg_embedding


if __name__ == "__main__":
    import pandas as pd
    df = pd.read_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\final_processed_reviews_deepl_use.csv", encoding="utf-8")
    df.dropna(subset=["hotel_name", "processed_final_review"], inplace=True)
    hotels = df["hotel_name"].unique()
    
    fixed_n_topics = 7  # Sabit topic sayısı
    results = []
    
    for hotel in hotels:
        hotel_reviews = df.loc[df["hotel_name"] == hotel, "processed_final_review"].tolist()
        if len(hotel_reviews) < 2:
            continue
        
        # 1) PCA
        avg_pca, var_ratios = pca_for_hotel(hotel_reviews)
        pca_pc1, pca_pc2 = float(avg_pca[0]), float(avg_pca[1])
        
        # 2) LDA -> presence
        lda_model, corpus, topic_dict, binary_lda = lda_for_hotel(hotel_reviews, fixed_n_topics)
        
        # 3) Topic-level sentiment
        # doc-topic distribution -> sentiment => -1,0,+1
        aspect_sentiments = compute_aspect_sentiment_for_hotel(
            lda_model, corpus, fixed_n_topics, hotel_reviews, threshold_doc=0.2
        )
        
        # 4) BERT embedding norm
        avg_bert = bert_for_hotel(hotel_reviews)
        bert_embedding_norm = float(np.linalg.norm(avg_bert))
        
        # Sonuç dict
        row_dict = {
            "hotel_name": hotel,
            "pca_pc1": pca_pc1,
            "pca_pc2": pca_pc2,
            "bert_embedding_norm": bert_embedding_norm
        }
        
        # Her topic için => presence, sentiment, keywords
        for i in range(fixed_n_topics):
            keywords = " ".join(topic_dict[i])  # LDA kelimeleri
            presence_val = int(binary_lda[i])   # 0/1
            sentiment_val = aspect_sentiments[i]  # -1,0,+1
            
            # Örneğin: topic_0_presence, topic_0_sentiment, topic_0_keywords
            row_dict[f"topic_{i}_presence"] = presence_val
            row_dict[f"topic_{i}_sentiment"] = sentiment_val
            row_dict[f"topic_{i}_keywords"] = keywords
        
        results.append(row_dict)
    
    # Son dataframe
    results_df = pd.DataFrame(results)
    results_df.to_csv("C:\\Users\\catsu\\PycharmProjects\\scrappyhotel\\hotel_feature_matrix_AS.csv",
                      index=False, encoding="utf-8-sig")
    
    print("CSV oluşturuldu: hotel_feature_matrix_AS.csv")
    print("Sütunlar:", results_df.columns.tolist())
```
